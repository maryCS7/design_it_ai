# DesignIt AI

A creative design platform using OpenAI's DALL·E and Gradio to turn text prompts into compelling visuals for marketing and storytelling.

```python
import os
import requests
from io import BytesIO
from PIL import Image
from dotenv import load_dotenv
from openai import OpenAI
import gradio as gr

load_dotenv()
client = OpenAI(api_key=os.getenv("OPENAI_API_KEY"))

# generate image with dalle
def generate_image(prompt):
    response = client.images.generate(
        model="dall-e-3",
        prompt=prompt,
        n=1,
        size="1024x1024" 
    )
    image_url = response.data[0].url
    image_response = requests.get(image_url)
    return Image.open(BytesIO(image_response.content))

# Gradio UI to turn user prompts into images
# chose `gr.Blocks` for gr interface for control over layout 
with gr.Blocks() as demo:
    gr.Markdown("## Creative Design Generator for Digital Campaigns")
    gr.Markdown(
        "Leverage AI to transform your marketing vision into bespoke visuals. "
        "Enter a description to generate high-quality imagery for banners, posters, and more."
    )
    with gr.Row(equal_height=True):
        textbox = gr.Textbox(
            placeholder="E.g. A futuristic city skyline at sunset, cinematic style",
            label="Design Prompt"
        )
        button = gr.Button("Generate Image", variant="primary")
    image_output = gr.Image(label="Generated Visual", height=384)

    # connect button to function
    button.click(fn=generate_image, inputs=textbox, outputs=image_output)

demo.launch()
```
